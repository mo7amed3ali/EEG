```python
# 1. Calculate Q1 and Q3 separately for each patient and channel
q1 = p2p_results.groupby(level='patient').transform(
    lambda x: x.quantile(0.25)
)

q3 = p2p_results.groupby(level='patient').transform(
    lambda x: x.quantile(0.75)
)

# 2. Calculate IQR
iqr = q3 - q1

# 3. Calculate upper and lower bounds
upper_bound = q3 + (1.5 * iqr)
lower_bound = q1 - (1.5 * iqr)

# 4. Detect potential outliers
outlier_flags = (
    (p2p_results > upper_bound) |
    (p2p_results < lower_bound)
)

# 5. Count outlier channels in each epoch
outlier_count = outlier_flags.sum(axis=1)

# 6. Create summary
print("Total Epochs:", len(p2p_results))
print("Epochs with potential outliers:", (outlier_count > 0).sum())
print("Epochs without potential outliers:", (outlier_count == 0).sum())

print("\nOutlier counts per patient:")
print(
    outlier_flags.groupby(level='patient')
    .sum()
)
```

```python
# Separate high and low outliers
high_outliers = p2p_results > upper_bound
low_outliers = p2p_results < lower_bound

# Count high outlier channels in each epoch
high_outlier_count = high_outliers.sum(axis=1)

# Count low outlier channels in each epoch
low_outlier_count = low_outliers.sum(axis=1)

# Distribution of high outlier counts
print("High Outlier Distribution:")
print(high_outlier_count.value_counts().sort_index())

print("\nEpochs with high outliers:")
print((high_outlier_count > 0).sum())

print("\nEpochs with multiple high outlier channels:")
print((high_outlier_count > 1).sum())

print("\nMaximum high outlier channels in one epoch:")
print(high_outlier_count.max())
```

```python
import numpy as np
import pandas as pd
from scipy.signal import welch
from scipy.integrate import trapezoid

fs = 250

bands = {
    'Theta': (4, 8),
    'Mu': (8, 13),
    'Beta': (13, 30)
}

band_power_before = []
band_power_after = []

for _, epoch_before in df_preprocessed.groupby(['patient', 'epoch']):
    epoch_after = df_filtered.loc[epoch_before.index]

    signals_before = epoch_before[eeg_channels].to_numpy()
    signals_after = epoch_after[eeg_channels].to_numpy()

    freqs, psd_before = welch(
        signals_before, fs=fs, nperseg=200, axis=0
    )

    _, psd_after = welch(
        signals_after, fs=fs, nperseg=200, axis=0
    )

    for band, (low, high) in bands.items():
        mask = (freqs >= low) & (freqs <= high)

        power_before = trapezoid(
            psd_before[mask], x=freqs[mask], axis=0
        )

        power_after = trapezoid(
            psd_after[mask], x=freqs[mask], axis=0
        )

        band_power_before.append({
            'Band': band,
            'Power': power_before.mean()
        })

        band_power_after.append({
            'Band': band,
            'Power': power_after.mean()
        })

before_df = pd.DataFrame(band_power_before)
after_df = pd.DataFrame(band_power_after)

before_summary = before_df.groupby('Band')['Power'].mean()
after_summary = after_df.groupby('Band')['Power'].mean()

band_comparison = pd.DataFrame({
    'Before Filtering': before_summary,
    'After Filtering': after_summary
})

band_comparison['Change %'] = (
    (band_comparison['After Filtering'] -
     band_comparison['Before Filtering'])
    / band_comparison['Before Filtering']
) * 100

display(band_comparison.round(4))
```

```python
import numpy as np
import pandas as pd
from scipy.signal import welch
from scipy.integrate import trapezoid

fs = 250

bands = {
    'Theta': (4, 8),
    'Mu': (8, 13),
    'Beta': (13, 30)
}

features = []

for (patient, epoch), group in df_filtered.groupby(['patient', 'epoch']):

    row = {
        'Patient': patient,
        'Epoch': epoch,
        'Label': group['label'].iloc[0]
    }

    for channel in eeg_channels:

        signal = group[channel].to_numpy()

        row[f'{channel}_Mean'] = np.mean(signal)
        row[f'{channel}_Std'] = np.std(signal)
        row[f'{channel}_Variance'] = np.var(signal)
        row[f'{channel}_RMS'] = np.sqrt(np.mean(signal ** 2))
        row[f'{channel}_P2P'] = np.ptp(signal)

        freqs, psd = welch(signal, fs=fs, nperseg=200)

        for band, (low, high) in bands.items():

            mask = (freqs >= low) & (freqs <= high)

            power = trapezoid(
                psd[mask],
                x=freqs[mask]
            )

            row[f'{channel}_{band}_Power'] = power

    features.append(row)

df_features = pd.DataFrame(features)

print("Features shape:", df_features.shape)
print("\nMissing values:", df_features.isnull().sum().sum())
print("\nClass distribution:")
print(df_features['Label'].value_counts())

df_features.head()
```

```python
import numpy as np
import pandas as pd

from mne.decoding import CSP
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score

# 1. Align frequency features with EEG epochs
epoch_keys = pd.MultiIndex.from_tuples(
    [(patient, epoch) for (patient, epoch), _ in epoch_groups],
    names=['Patient', 'Epoch']
)

frequency_df = (
    df_features
    .set_index(['Patient', 'Epoch'])
    .reindex(epoch_keys)
)

X_frequency = frequency_df[frequency_features].to_numpy()
y_frequency = frequency_df['Label'].to_numpy()

# Verify alignment
assert not np.isnan(X_frequency).any()
assert np.array_equal(y_frequency, y_epochs)

print("Frequency shape:", X_frequency.shape)
print("Epoch shape:", X_epochs.shape)
print("Labels aligned:", np.array_equal(y_frequency, y_epochs))


# 2. LOSO Cross-Validation
logo = LeaveOneGroupOut()

models_results = {
    'Frequency Only': [],
    'CSP Only': [],
    'CSP + Frequency': []
}

for train_idx, test_idx in logo.split(
    X_epochs, y_epochs, groups_epochs
):

    # Fit CSP using training data only
    csp = CSP(
        n_components=8,
        reg='ledoit_wolf',
        log=True,
        norm_trace=False
    )

    X_csp_train = csp.fit_transform(
        X_epochs[train_idx],
        y_epochs[train_idx]
    )

    X_csp_test = csp.transform(X_epochs[test_idx])

    # Prepare the three feature configurations
    datasets = {
        'Frequency Only': (
            X_frequency[train_idx],
            X_frequency[test_idx]
        ),

        'CSP Only': (
            X_csp_train,
            X_csp_test
        ),

        'CSP + Frequency': (
            np.hstack([
                X_csp_train,
                X_frequency[train_idx]
            ]),
            np.hstack([
                X_csp_test,
                X_frequency[test_idx]
            ])
        )
    }

    # Train and evaluate each configuration
    for name, (X_train_fold, X_test_fold) in datasets.items():

        scaler = StandardScaler()

        X_train_scaled = scaler.fit_transform(X_train_fold)
        X_test_scaled = scaler.transform(X_test_fold)

        clf = LogisticRegression(
            max_iter=2000,
            random_state=42
        )

        clf.fit(X_train_scaled, y_epochs[train_idx])

        y_pred = clf.predict(X_test_scaled)

        models_results[name].append({
            'Patient': groups_epochs[test_idx][0],
            'Accuracy': accuracy_score(
                y_epochs[test_idx], y_pred
            ),
            'Macro F1': f1_score(
                y_epochs[test_idx],
                y_pred,
                average='macro',
                zero_division=0
            )
        })


# 3. Display patient-wise results
for name, results in models_results.items():

    print(f"\n{'=' * 45}")
    print(name)
    print('=' * 45)

    results_df = pd.DataFrame(results)

    print(results_df.round(4).to_string(index=False))


# 4. Compare overall performance
summary = []

for name, results in models_results.items():

    results_df = pd.DataFrame(results)

    summary.append({
        'Model': name,
        'Mean Accuracy': results_df['Accuracy'].mean(),
        'Std Accuracy': results_df['Accuracy'].std(ddof=0),
        'Mean Macro F1': results_df['Macro F1'].mean(),
        'Std Macro F1': results_df['Macro F1'].std(ddof=0)
    })

summary_df = pd.DataFrame(summary)

print("\nFinal Comparison:")
print(summary_df.round(4).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import (
    LeaveOneGroupOut,
    StratifiedGroupKFold,
    GridSearchCV
)

from sklearn.metrics import accuracy_score, f1_score

# Outer CV: Leave One Patient Out
logo = LeaveOneGroupOut()

# Hyperparameter search space
param_grid = {
    'classifier__kernel': ['linear', 'rbf'],
    'classifier__C': [0.1, 1, 10, 100],
    'classifier__gamma': ['scale', 'auto'],
    'classifier__class_weight': [None, 'balanced']
}

svm_results = []
best_parameters = []

# Nested LOSO
for fold, (train_idx, test_idx) in enumerate(
    logo.split(X_frequency, y_epochs, groups_epochs), 1
):

    X_train_fold = X_frequency[train_idx]
    X_test_fold = X_frequency[test_idx]

    y_train_fold = y_epochs[train_idx]
    y_test_fold = y_epochs[test_idx]

    groups_train_fold = groups_epochs[train_idx]

    # Pipeline prevents scaling leakage
    svm_pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('classifier', SVC())
    ])

    # Inner CV: patient-aware validation
    inner_cv = StratifiedGroupKFold(
        n_splits=4,
        shuffle=True,
        random_state=42
    )

    grid_search = GridSearchCV(
        estimator=svm_pipeline,
        param_grid=param_grid,
        scoring='f1_macro',
        cv=inner_cv,
        n_jobs=-1,
        refit=True
    )

    # Tune using training patients only
    grid_search.fit(
        X_train_fold,
        y_train_fold,
        groups=groups_train_fold
    )

    # Evaluate on the unseen patient
    best_model = grid_search.best_estimator_
    y_pred = best_model.predict(X_test_fold)

    patient = groups_epochs[test_idx][0]

    accuracy = accuracy_score(y_test_fold, y_pred)
    macro_f1 = f1_score(
        y_test_fold,
        y_pred,
        average='macro',
        zero_division=0
    )

    svm_results.append({
        'Patient': patient,
        'Accuracy': accuracy,
        'Macro F1': macro_f1
    })

    best_parameters.append({
        'Patient': patient,
        'Best Parameters': grid_search.best_params_,
        'Inner CV F1': grid_search.best_score_
    })

    print(f"Fold {fold}/9 | Patient {patient}")
    print(f"Best Parameters: {grid_search.best_params_}")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Macro F1: {macro_f1:.4f}")
    print("-" * 50)


# Patient-wise results
svm_results_df = pd.DataFrame(svm_results)

print("\nPatient-wise Results:")
print(svm_results_df.round(4).to_string(index=False))


# Overall performance
print("\nFinal Nested LOSO Results:")
print(f"Mean Accuracy: {svm_results_df['Accuracy'].mean():.4f}")
print(f"Std Accuracy: {svm_results_df['Accuracy'].std(ddof=0):.4f}")

print(f"Mean Macro F1: {svm_results_df['Macro F1'].mean():.4f}")
print(f"Std Macro F1: {svm_results_df['Macro F1'].std(ddof=0):.4f}")
```

```python
from scipy.signal import welch
from scipy.integrate import trapezoid
import pandas as pd
import numpy as np

fs = 250

bands = {
    'Theta': (4, 8),
    'Mu': (8, 13),
    'Beta': (13, 30)
}

new_features = {}

for channel_idx, channel in enumerate(eeg_channels):

    signal = X_new_epoch[channel_idx]

    freqs, psd = welch(signal, fs=fs, nperseg=200)

    for band, (low, high) in bands.items():

        mask = (freqs >= low) & (freqs <= high)

        power = trapezoid(psd[mask], x=freqs[mask])

        new_features[f'{channel}_{band}_Power'] = power

X_new_features = pd.DataFrame([new_features])

# Match the exact training feature order
X_new_features = X_new_features[frequency_features]

print("Extracted Features Shape:", X_new_features.shape)
print("Number of Features:", len(X_new_features.columns))
```
